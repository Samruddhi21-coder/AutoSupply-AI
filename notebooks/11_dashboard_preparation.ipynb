```python
# ============================================================
# AUTO SUPPLY AI
# NOTEBOOK 11 — DASHBOARD PREPARATION
# CELL 1: IMPORTS & PROJECT SETUP
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# PROJECT PATHS
# ------------------------------------------------------------

PROJECT_DIR = Path.cwd().parent
NOTEBOOKS_DIR = PROJECT_DIR / "notebooks"
DATA_DIR = PROJECT_DIR / "data"
PROCESSED_DIR = DATA_DIR / "processed"
RAW_DIR = DATA_DIR / "raw"
MODELS_DIR = PROJECT_DIR / "models"

# ------------------------------------------------------------
# DISPLAY SETTINGS
# ------------------------------------------------------------

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
pd.set_option("display.float_format", "{:,.2f}".format)

print("=" * 75)
print("AUTO SUPPLY AI — DASHBOARD PREPARATION")
print("=" * 75)

print("\nProject Directory:")
print(PROJECT_DIR)

print("\nNotebooks Directory:")
print(NOTEBOOKS_DIR)

print("\nProcessed Data Directory:")
print(PROCESSED_DIR)

print("\nSetup completed successfully.")
print("=" * 75)
```

```python
# ============================================================
# CELL 2: LOAD DASHBOARD DATASETS
# ============================================================

print("=" * 75)
print("LOADING DASHBOARD DATASETS")
print("=" * 75)


def load_notebook_csv(filename):
    file_path = NOTEBOOKS_DIR / filename

    if not file_path.exists():
        raise FileNotFoundError(
            f"\nCSV file not found:\n{file_path}"
        )

    df = pd.read_csv(file_path)

    print(f"✓ {filename:<42} Shape: {df.shape}")

    return df


# ------------------------------------------------------------
# 1. FINAL DEMAND FORECAST
# ------------------------------------------------------------

demand_forecast_df = load_notebook_csv(
    "final_demand_forecast.csv"
)


# ------------------------------------------------------------
# 2. INVENTORY OPTIMIZATION
# ------------------------------------------------------------

inventory_df = load_notebook_csv(
    "inventory_optimization.csv"
)


# ------------------------------------------------------------
# 3. STATE ANALYSIS
# ------------------------------------------------------------

state_analysis_df = load_notebook_csv(
    "autosupply_state_analysis.csv"
)


# ------------------------------------------------------------
# 4. VEHICLE GROUP ANALYSIS
# ------------------------------------------------------------

vehicle_analysis_df = load_notebook_csv(
    "autosupply_vehicle_analysis.csv"
)


# ------------------------------------------------------------
# 5. STATE-VEHICLE ANALYSIS
# ------------------------------------------------------------

state_vehicle_analysis_df = load_notebook_csv(
    "autosupply_state_vehicle_analysis.csv"
)


# ------------------------------------------------------------
# 6. MONTHLY FORECAST
# ------------------------------------------------------------

monthly_forecast_df = load_notebook_csv(
    "autosupply_monthly_forecast.csv"
)


# ------------------------------------------------------------
# 7. CREATE SUPPLY DATASET FROM INVENTORY DATA
# ------------------------------------------------------------

supply_df = inventory_df.copy()

# Supply requirement = Forecast Demand + Safety Stock
if "total_supply_requirement" not in supply_df.columns:

    supply_df["total_supply_requirement"] = (
        supply_df["total_forecast_demand"]
        + supply_df["safety_stock"]
    )

print("\n✓ Supply requirement dataset prepared from inventory data")


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ALL DASHBOARD DATASETS LOADED SUCCESSFULLY")
print("=" * 75)

print("\nDataset Summary:")

print(f"Demand Forecast       : {demand_forecast_df.shape}")
print(f"Inventory Optimization: {inventory_df.shape}")
print(f"Supply Requirement    : {supply_df.shape}")
print(f"State Analysis        : {state_analysis_df.shape}")
print(f"Vehicle Analysis      : {vehicle_analysis_df.shape}")
print(f"State-Vehicle Analysis: {state_vehicle_analysis_df.shape}")
print(f"Monthly Forecast      : {monthly_forecast_df.shape}")

print("\n" + "=" * 75)
```
