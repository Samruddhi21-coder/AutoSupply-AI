```python
# ============================================================
# AUTO SUPPLY AI — DASHBOARD DATA PREPARATION
# Cell 1: Setup & File Discovery
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Project directories
# ------------------------------------------------------------

PROJECT_DIR = Path.cwd().parent
NOTEBOOKS_DIR = PROJECT_DIR / "notebooks"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
MODELS_DIR = PROJECT_DIR / "models"

print("=" * 75)
print("AUTO SUPPLY AI — DASHBOARD DATA PREPARATION")
print("=" * 75)

print(f"\nProject Directory:")
print(PROJECT_DIR)

print(f"\nNotebooks Directory:")
print(NOTEBOOKS_DIR)

print(f"\nProcessed Data Directory:")
print(PROCESSED_DIR)

# ------------------------------------------------------------
# Check directories
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("DIRECTORY CHECK")
print("-" * 75)

print(f"✓ Project directory: {PROJECT_DIR.exists()}")
print(f"✓ Notebooks directory: {NOTEBOOKS_DIR.exists()}")
print(f"✓ Processed directory: {PROCESSED_DIR.exists()}")

# ------------------------------------------------------------
# Find CSV files currently available
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("CSV FILES AVAILABLE IN NOTEBOOKS FOLDER")
print("-" * 75)

notebook_csvs = sorted(NOTEBOOKS_DIR.glob("*.csv"))

print(f"\nFound {len(notebook_csvs)} CSV file(s):")

for file in notebook_csvs:
    print(f"  ✓ {file.name}")

# ------------------------------------------------------------
# Find CSV files in processed folder
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("CSV FILES AVAILABLE IN PROCESSED FOLDER")
print("-" * 75)

processed_csvs = sorted(PROCESSED_DIR.glob("*.csv"))

print(f"\nFound {len(processed_csvs)} CSV file(s):")

for file in processed_csvs:
    print(f"  ✓ {file.name}")

print("\n" + "=" * 75)
print("SETUP COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# Cell 2: Load Dashboard Source Data
# ============================================================

def load_notebook_csv(filename):
    """
    Load a CSV from the notebooks folder.
    """
    file_path = NOTEBOOKS_DIR / filename

    if not file_path.exists():
        raise FileNotFoundError(
            f"CSV file not found:\n{file_path}"
        )

    df = pd.read_csv(file_path)

    print(f"✓ Loaded: {filename}")
    print(f"  Shape: {df.shape}")

    return df


# ------------------------------------------------------------
# Load required dashboard datasets
# ------------------------------------------------------------

inventory_df = load_notebook_csv(
    "inventory_optimization.csv"
)

monthly_forecast_df = load_notebook_csv(
    "autosupply_monthly_forecast.csv"
)

state_analysis_df = load_notebook_csv(
    "autosupply_state_analysis.csv"
)

vehicle_analysis_df = load_notebook_csv(
    "autosupply_vehicle_analysis.csv"
)

state_vehicle_analysis_df = load_notebook_csv(
    "autosupply_state_vehicle_analysis.csv"
)


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DASHBOARD DATA LOADED SUCCESSFULLY")
print("=" * 75)

print(f"\nInventory data:       {inventory_df.shape}")
print(f"Monthly forecast:     {monthly_forecast_df.shape}")
print(f"State analysis:       {state_analysis_df.shape}")
print(f"Vehicle analysis:     {vehicle_analysis_df.shape}")
print(f"State-vehicle data:   {state_vehicle_analysis_df.shape}")
```

```python
# ============================================================
# Cell 3: Inspect Dashboard Data
# ============================================================

print("=" * 75)
print("DASHBOARD DATASET INSPECTION")
print("=" * 75)

datasets = {
    "Inventory": inventory_df,
    "Monthly Forecast": monthly_forecast_df,
    "State Analysis": state_analysis_df,
    "Vehicle Analysis": vehicle_analysis_df,
    "State × Vehicle": state_vehicle_analysis_df
}

for name, df in datasets.items():

    print("\n" + "-" * 75)
    print(name)
    print("-" * 75)

    print("Shape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nMissing values:")
    print(df.isna().sum().to_dict())

    print("\nFirst 3 rows:")
    display(df.head(3))

print("\n" + "=" * 75)
print("INSPECTION COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# Cell 4: Dashboard KPI Calculations
# ============================================================

print("=" * 75)
print("AUTOSUPPLY-AI — DASHBOARD KPIs")
print("=" * 75)

# ------------------------------------------------------------
# Basic counts
# ------------------------------------------------------------

total_states = inventory_df["state"].nunique()
total_vehicle_groups = inventory_df["vehicle_group"].nunique()

# ------------------------------------------------------------
# Forecast KPIs
# ------------------------------------------------------------

total_forecast_demand = inventory_df["total_forecast_demand"].sum()

average_monthly_demand = inventory_df["average_monthly_demand"].mean()

peak_forecast_demand = inventory_df["total_forecast_demand"].max()

forecast_months = inventory_df["forecast_months"].max()

# ------------------------------------------------------------
# Peak demand combination
# ------------------------------------------------------------

peak_row = inventory_df.loc[
    inventory_df["total_forecast_demand"].idxmax()
]

peak_state = peak_row["state"]
peak_vehicle = peak_row["vehicle_group"]

# ------------------------------------------------------------
# Inventory KPIs
# ------------------------------------------------------------

total_safety_stock = inventory_df["safety_stock"].sum()

total_recommended_inventory = (
    inventory_df["recommended_inventory"].sum()
)

high_priority_count = (
    inventory_df["inventory_priority"]
    .eq("HIGH")
    .sum()
)

medium_priority_count = (
    inventory_df["inventory_priority"]
    .eq("MEDIUM")
    .sum()
)

low_priority_count = (
    inventory_df["inventory_priority"]
    .eq("LOW")
    .sum()
)

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("\nKEY PERFORMANCE INDICATORS")
print("-" * 75)

print(f"Total Forecasted Demand      : {total_forecast_demand:,.2f}")
print(f"Average Monthly Demand       : {average_monthly_demand:,.2f}")
print(f"Peak Forecast Demand         : {peak_forecast_demand:,.2f}")
print(f"Forecast Months              : {forecast_months}")
print(f"States                       : {total_states}")
print(f"Vehicle Groups               : {total_vehicle_groups}")

print("\nPEAK DEMAND COMBINATION")
print("-" * 75)

print(f"State                        : {peak_state}")
print(f"Vehicle Group                : {peak_vehicle}")
print(f"Total Forecast Demand        : {peak_forecast_demand:,.2f}")

print("\nINVENTORY KPIs")
print("-" * 75)

print(f"Total Safety Stock           : {total_safety_stock:,.2f}")
print(f"Recommended Inventory        : {total_recommended_inventory:,.2f}")

print(f"HIGH Priority Combinations   : {high_priority_count}")
print(f"MEDIUM Priority Combinations : {medium_priority_count}")
print(f"LOW Priority Combinations    : {low_priority_count}")

print("\n" + "=" * 75)
print("KPI CALCULATION COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# Cell 5: Check Inventory Columns
# ============================================================

print("=" * 75)
print("INVENTORY DATA — ACTUAL COLUMNS")
print("=" * 75)

print("\nColumns in inventory_df:")
for i, col in enumerate(inventory_df.columns):
    print(f"{i}: {col}")

print("\n" + "-" * 75)
print("FIRST 5 ROWS")
print("-" * 75)

display(inventory_df.head())

print("\n" + "=" * 75)
print("COLUMN CHECK COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# CELL 5 — VEHICLE GROUP DASHBOARD ANALYSIS
# ============================================================

print("=" * 75)
print("VEHICLE GROUP DEMAND & INVENTORY ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# Aggregate by vehicle group
# ------------------------------------------------------------

vehicle_kpi = (
    inventory_df
    .groupby("vehicle_group", as_index=False)
    .agg(
        total_forecast_demand=("total_forecast_demand", "sum"),
        average_monthly_demand=("average_monthly_demand", "mean"),
        total_safety_stock=("safety_stock", "sum"),
        total_recommended_inventory=("recommended_inventory", "sum")
    )
)

# ------------------------------------------------------------
# Calculate demand share
# ------------------------------------------------------------

vehicle_kpi["demand_share_percent"] = (
    vehicle_kpi["total_forecast_demand"]
    / vehicle_kpi["total_forecast_demand"].sum()
    * 100
)

# ------------------------------------------------------------
# Sort by total forecast demand
# ------------------------------------------------------------

vehicle_kpi = (
    vehicle_kpi
    .sort_values(
        "total_forecast_demand",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("\nVEHICLE GROUP SUMMARY")
print("-" * 75)

display(vehicle_kpi)

print("\n" + "=" * 75)
print("VEHICLE GROUP ANALYSIS COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# CELL 6 — STATE KPI ANALYSIS
# ============================================================

print("=" * 75)
print("STATE DEMAND & INVENTORY ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# Aggregate inventory data by state
# ------------------------------------------------------------

state_kpi = (
    inventory_df
    .groupby("state", as_index=False)
    .agg(
        total_forecast_demand=("total_forecast_demand", "sum"),
        average_monthly_demand=("average_monthly_demand", "sum"),
        total_safety_stock=("safety_stock", "sum"),
        total_recommended_inventory=("recommended_inventory", "sum")
    )
)

# ------------------------------------------------------------
# Calculate demand share
# ------------------------------------------------------------

state_kpi["demand_share_percent"] = (
    state_kpi["total_forecast_demand"]
    / state_kpi["total_forecast_demand"].sum()
    * 100
)

# ------------------------------------------------------------
# Sort states by forecast demand
# ------------------------------------------------------------

state_kpi = (
    state_kpi
    .sort_values(
        "total_forecast_demand",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Display top 10 states
# ------------------------------------------------------------

print("\nTOP 10 STATES BY FORECAST DEMAND")
print("-" * 75)

display(state_kpi.head(10))

print("\n" + "=" * 75)
print("STATE KPI ANALYSIS COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# TOP 10 STATE-VEHICLE COMBINATIONS
# ============================================================

print("=" * 75)
print("TOP 10 STATE-VEHICLE COMBINATIONS BY FORECAST DEMAND")
print("=" * 75)

top_state_vehicle = (
    inventory_df
    .groupby(["state", "vehicle_group"], as_index=False)
    .agg(
        total_forecast_demand=("total_forecast_demand", "sum"),
        average_monthly_demand=("average_monthly_demand", "mean"),
        safety_stock=("safety_stock", "sum"),
        recommended_inventory=("recommended_inventory", "sum")
    )
    .sort_values("total_forecast_demand", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

display(top_state_vehicle)
```

```python
# ============================================================
# STATE DEMAND SHARE
# ============================================================

print("=" * 75)
print("STATE DEMAND SHARE")
print("=" * 75)

state_demand_share = (
    inventory_df
    .groupby("state", as_index=False)
    .agg(
        total_forecast_demand=("total_forecast_demand", "sum"),
        average_monthly_demand=("average_monthly_demand", "mean"),
        total_safety_stock=("safety_stock", "sum"),
        total_recommended_inventory=("recommended_inventory", "sum")
    )
)

total_demand = state_demand_share["total_forecast_demand"].sum()

state_demand_share["demand_share_percent"] = (
    state_demand_share["total_forecast_demand"] / total_demand * 100
)

state_demand_share = (
    state_demand_share
    .sort_values("total_forecast_demand", ascending=False)
    .reset_index(drop=True)
)

display(state_demand_share.head(10))
```

```python
# ============================================================
# INVENTORY PRIORITY SUMMARY
# ============================================================

print("=" * 75)
print("INVENTORY PRIORITY SUMMARY")
print("=" * 75)

priority_summary = (
    inventory_df
    .groupby("inventory_priority", as_index=False)
    .agg(
        count=("inventory_priority", "size"),
        total_forecast_demand=("total_forecast_demand", "sum"),
        total_safety_stock=("safety_stock", "sum"),
        total_recommended_inventory=("recommended_inventory", "sum")
    )
)

priority_summary = priority_summary.sort_values(
    "total_forecast_demand",
    ascending=False
).reset_index(drop=True)

display(priority_summary)
```

```python
# ============================================================
# DASHBOARD SUMMARY DATA
# ============================================================

print("=" * 75)
print("DASHBOARD SUMMARY DATA")
print("=" * 75)

dashboard_summary = {
    "total_forecast_demand": inventory_df["total_forecast_demand"].sum(),
    "total_safety_stock": inventory_df["safety_stock"].sum(),
    "total_recommended_inventory": inventory_df["recommended_inventory"].sum(),
    "total_state_vehicle_combinations": len(inventory_df),
    "total_states": inventory_df["state"].nunique(),
    "total_vehicle_groups": inventory_df["vehicle_group"].nunique()
}

for key, value in dashboard_summary.items():
    print(f"{key}: {value:,.2f}" if isinstance(value, (int, float)) else f"{key}: {value}")
```

```python
# ============================================================
# CELL 11 — SAVE DASHBOARD DATASETS
# ============================================================

print("=" * 75)
print("SAVING DASHBOARD DATASETS")
print("=" * 75)

# Save all dashboard-ready datasets
vehicle_kpi.to_csv("vehicle_kpi_dashboard.csv", index=False)
state_kpi.to_csv("state_kpi_dashboard.csv", index=False)
top_state_vehicle.to_csv("top_state_vehicle_dashboard.csv", index=False)
state_demand_share.to_csv("state_demand_share_dashboard.csv", index=False)
priority_summary.to_csv("priority_summary_dashboard.csv", index=False)

print("\nFiles created successfully:")
print("1. vehicle_kpi_dashboard.csv")
print("2. state_kpi_dashboard.csv")
print("3. top_state_vehicle_dashboard.csv")
print("4. state_demand_share_dashboard.csv")
print("5. priority_summary_dashboard.csv")

print("\n" + "=" * 75)
print("DASHBOARD DATASETS SAVED")
print("=" * 75)
```

```python
# ============================================================
# CELL 12 — FINAL DASHBOARD DATA VALIDATION
# ============================================================

print("=" * 75)
print("FINAL DASHBOARD DATA VALIDATION")
print("=" * 75)

# ------------------------------------------------------------
# Check all dashboard datasets
# ------------------------------------------------------------

dashboard_files = {
    "vehicle_kpi": vehicle_kpi,
    "state_kpi": state_kpi,
    "top_state_vehicle": top_state_vehicle,
    "state_demand_share": state_demand_share,
    "priority_summary": priority_summary
}

for name, df in dashboard_files.items():

    print(f"\n{name}")
    print("-" * 75)

    print(f"Rows    : {len(df)}")
    print(f"Columns : {len(df.columns)}")
    print(f"Missing : {df.isna().sum().sum()}")
    print(f"Duplicates : {df.duplicated().sum()}")

# ------------------------------------------------------------
# Validate total forecast demand
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FORECAST TOTAL VALIDATION")
print("=" * 75)

inventory_total = inventory_df["total_forecast_demand"].sum()
state_total = state_kpi["total_forecast_demand"].sum()
vehicle_total = vehicle_kpi["total_forecast_demand"].sum()

print(f"Inventory total : {inventory_total:,.2f}")
print(f"State KPI total : {state_total:,.2f}")
print(f"Vehicle KPI total: {vehicle_total:,.2f}")

# ------------------------------------------------------------
# Final status
# ------------------------------------------------------------

if abs(inventory_total - state_total) < 0.01 and \
   abs(inventory_total - vehicle_total) < 0.01:

    print("\nSTATUS: PASS")
    print("All dashboard forecast totals are consistent.")

else:

    print("\nSTATUS: CHECK REQUIRED")
    print("Dashboard totals do not match.")

print("\n" + "=" * 75)
print("DASHBOARD DATA VALIDATION COMPLETE")
print("=" * 75)
```
