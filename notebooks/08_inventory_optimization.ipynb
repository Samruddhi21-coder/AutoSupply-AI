```python
# ============================================================
# AUTOSUPPLY-AI - INVENTORY OPTIMIZATION
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# PROJECT PATH
# ------------------------------------------------------------

PROJECT_ROOT = Path(r"F:\AutoSupply-AI")

# IMPORTANT:
# final_demand_forecast.csv is located inside notebooks/
FORECAST_FILE = PROJECT_ROOT / "notebooks" / "final_demand_forecast.csv"

print("=" * 70)
print("AUTOSUPPLY-AI - INVENTORY OPTIMIZATION")
print("=" * 70)

print("\nProject Root:")
print(PROJECT_ROOT)

print("\nForecast File:")
print(FORECAST_FILE)

# ------------------------------------------------------------
# CHECK FILE
# ------------------------------------------------------------

if not FORECAST_FILE.exists():
    raise FileNotFoundError(
        f"\nForecast file not found:\n{FORECAST_FILE}"
    )

print("\n✓ Forecast file found!")

# ------------------------------------------------------------
# LOAD FORECAST DATA
# ------------------------------------------------------------

forecast_df = pd.read_csv(FORECAST_FILE)

print("\n✓ Forecast dataset loaded successfully!")

print("\nDataset Shape:")
print(forecast_df.shape)

print("\nColumns:")
print(forecast_df.columns.tolist())

# ------------------------------------------------------------
# BASIC VALIDATION
# ------------------------------------------------------------

required_columns = [
    "month",
    "state",
    "vehicle_group",
    "predicted_demand"
]

missing_columns = [
    col for col in required_columns
    if col not in forecast_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# Convert month to datetime
forecast_df["month"] = pd.to_datetime(
    forecast_df["month"]
)

# Ensure demand is numeric
forecast_df["predicted_demand"] = pd.to_numeric(
    forecast_df["predicted_demand"],
    errors="coerce"
)

# Check missing values
print("\nMissing Values:")
print(forecast_df[required_columns].isnull().sum())

# ------------------------------------------------------------
# DISPLAY DATA
# ------------------------------------------------------------

print("\nFirst 10 rows:")
display(forecast_df.head(10))

print("\nForecast Summary:")
print(forecast_df[required_columns].describe(include="all"))

print("\n" + "=" * 70)
print("FORECAST DATA LOADED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# INVENTORY OPTIMIZATION - DEMAND SUMMARY
# ============================================================

print("=" * 70)
print("INVENTORY DEMAND SUMMARY")
print("=" * 70)

# ------------------------------------------------------------
# 1. TOTAL DEMAND BY STATE
# ------------------------------------------------------------

state_demand = (
    forecast_df
    .groupby("state", as_index=False)["predicted_demand"]
    .sum()
    .sort_values("predicted_demand", ascending=False)
)

print("\nTop 10 States by Forecasted Demand:")
display(state_demand.head(10))

# ------------------------------------------------------------
# 2. TOTAL DEMAND BY VEHICLE GROUP
# ------------------------------------------------------------

vehicle_demand = (
    forecast_df
    .groupby("vehicle_group", as_index=False)["predicted_demand"]
    .sum()
    .sort_values("predicted_demand", ascending=False)
)

print("\nDemand by Vehicle Group:")
display(vehicle_demand)

# ------------------------------------------------------------
# 3. MONTHLY DEMAND
# ------------------------------------------------------------

monthly_demand = (
    forecast_df
    .groupby("month", as_index=False)["predicted_demand"]
    .sum()
    .sort_values("month")
)

print("\nMonthly Forecasted Demand:")
display(monthly_demand)

# ------------------------------------------------------------
# 4. OVERALL DEMAND
# ------------------------------------------------------------

total_demand = forecast_df["predicted_demand"].sum()
average_demand = forecast_df["predicted_demand"].mean()

print("\n" + "-" * 70)
print(f"Total Forecasted Demand : {total_demand:,.2f}")
print(f"Average Prediction      : {average_demand:,.2f}")
print(f"States                  : {forecast_df['state'].nunique()}")
print(f"Vehicle Groups          : {forecast_df['vehicle_group'].nunique()}")
print(f"Forecast Months         : {forecast_df['month'].nunique()}")
print("-" * 70)

# ------------------------------------------------------------
# 5. PEAK DEMAND
# ------------------------------------------------------------

peak_row = forecast_df.loc[
    forecast_df["predicted_demand"].idxmax()
]

print("\nPeak Demand Combination:")
print(f"State          : {peak_row['state']}")
print(f"Vehicle Group  : {peak_row['vehicle_group']}")
print(f"Month          : {peak_row['month'].strftime('%Y-%m-%d')}")
print(f"Demand         : {peak_row['predicted_demand']:,.2f}")

print("\n" + "=" * 70)
print("DEMAND SUMMARY COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# AUTOSUPPLY-AI - INVENTORY OPTIMIZATION
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. PROJECT PATHS
# ------------------------------------------------------------

PROJECT_ROOT = Path(r"F:\AutoSupply-AI")

# IMPORTANT:
# final_demand_forecast.csv is inside the notebooks folder
FORECAST_FILE = PROJECT_ROOT / "notebooks" / "final_demand_forecast.csv"

# Output will also be saved inside notebooks
OUTPUT_FILE = PROJECT_ROOT / "notebooks" / "inventory_optimization.csv"

print("=" * 70)
print("AUTOSUPPLY-AI - INVENTORY OPTIMIZATION")
print("=" * 70)

print("\nProject Root:")
print(PROJECT_ROOT)

print("\nForecast File:")
print(FORECAST_FILE)

# ------------------------------------------------------------
# 2. CHECK FILE
# ------------------------------------------------------------

if not FORECAST_FILE.exists():
    raise FileNotFoundError(
        f"\nForecast file not found:\n{FORECAST_FILE}"
    )

print("\n✓ Forecast file found!")

# ------------------------------------------------------------
# 3. LOAD FORECAST DATA
# ------------------------------------------------------------

forecast_df = pd.read_csv(FORECAST_FILE)

print("\nForecast data loaded successfully!")
print("Shape:", forecast_df.shape)

print("\nColumns:")
print(forecast_df.columns.tolist())

print("\nFirst 5 rows:")
display(forecast_df.head())
```

```python
# ============================================================
# 4. INVENTORY OPTIMIZATION
# ============================================================

print("=" * 70)
print("INVENTORY OPTIMIZATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. PREPARE DATA
# ------------------------------------------------------------

forecast_df["month"] = pd.to_datetime(forecast_df["month"])

forecast_df["predicted_demand"] = pd.to_numeric(
    forecast_df["predicted_demand"],
    errors="coerce"
)

# Remove invalid demand values
forecast_df = forecast_df.dropna(
    subset=["state", "vehicle_group", "month", "predicted_demand"]
).copy()

# Make sure demand cannot be negative
forecast_df["predicted_demand"] = forecast_df[
    "predicted_demand"
].clip(lower=0)

# ------------------------------------------------------------
# 2. INVENTORY ASSUMPTIONS
# ------------------------------------------------------------

# Assume inventory should cover approximately 1 month
LEAD_TIME_MONTHS = 1

# 95% service level
# Z-score for approximately 95% service level
SERVICE_LEVEL_Z = 1.65

print("\nInventory Assumptions:")
print("Lead Time:", LEAD_TIME_MONTHS, "month")
print("Service Level: 95%")
print("Safety Stock Z-score:", SERVICE_LEVEL_Z)

# ------------------------------------------------------------
# 3. CALCULATE DEMAND STATISTICS
# ------------------------------------------------------------

inventory_df = (
    forecast_df
    .groupby(["state", "vehicle_group"])
    .agg(
        average_monthly_demand=("predicted_demand", "mean"),
        demand_std=("predicted_demand", "std"),
        total_forecast_demand=("predicted_demand", "sum"),
        forecast_months=("predicted_demand", "count")
    )
    .reset_index()
)

# If only one observation exists, standard deviation becomes NaN
inventory_df["demand_std"] = inventory_df["demand_std"].fillna(0)

# ------------------------------------------------------------
# 4. CALCULATE SAFETY STOCK
# ------------------------------------------------------------

inventory_df["safety_stock"] = (
    SERVICE_LEVEL_Z
    * inventory_df["demand_std"]
    * np.sqrt(LEAD_TIME_MONTHS)
)

# ------------------------------------------------------------
# 5. CALCULATE REORDER POINT
# ------------------------------------------------------------

inventory_df["reorder_point"] = (
    inventory_df["average_monthly_demand"]
    * LEAD_TIME_MONTHS
    + inventory_df["safety_stock"]
)

# ------------------------------------------------------------
# 6. RECOMMENDED INVENTORY LEVEL
# ------------------------------------------------------------

inventory_df["recommended_inventory"] = (
    inventory_df["reorder_point"]
).clip(lower=0)

# Round values
numeric_columns = [
    "average_monthly_demand",
    "demand_std",
    "total_forecast_demand",
    "safety_stock",
    "reorder_point",
    "recommended_inventory"
]

inventory_df[numeric_columns] = inventory_df[numeric_columns].round(2)

# ------------------------------------------------------------
# 7. INVENTORY PRIORITY
# ------------------------------------------------------------

inventory_df["inventory_priority"] = pd.cut(
    inventory_df["average_monthly_demand"],
    bins=[-np.inf, 1000, 10000, np.inf],
    labels=["LOW", "MEDIUM", "HIGH"]
)

# ------------------------------------------------------------
# 8. SORT BY DEMAND
# ------------------------------------------------------------

inventory_df = inventory_df.sort_values(
    "average_monthly_demand",
    ascending=False
).reset_index(drop=True)

# ------------------------------------------------------------
# 9. DISPLAY RESULT
# ------------------------------------------------------------

print("\nInventory Optimization Results:")
print("Number of State-Vehicle Combinations:", len(inventory_df))

display(inventory_df.head(20))

print("\n" + "=" * 70)
print("INVENTORY OPTIMIZATION CALCULATION COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# 15. SAVE INVENTORY OPTIMIZATION RESULTS
# ============================================================

# Output file location
OUTPUT_FILE = PROJECT_ROOT / "notebooks" / "inventory_optimization.csv"

# Save results
inventory_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n" + "=" * 70)
print("INVENTORY OPTIMIZATION COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nOutput File:")
print(OUTPUT_FILE)

print("\nRows Saved:", len(inventory_df))
print("Columns Saved:", len(inventory_df.columns))

print("\nSaved Columns:")
print(inventory_df.columns.tolist())

print("\n✓ File saved successfully!")
```

```python
# ============================================================
# 16. FINAL VALIDATION
# ============================================================

saved_inventory = pd.read_csv(OUTPUT_FILE)

print("=" * 70)
print("FINAL INVENTORY OUTPUT VALIDATION")
print("=" * 70)

print("\nFile Exists:", OUTPUT_FILE.exists())
print("Rows:", len(saved_inventory))
print("Columns:", len(saved_inventory.columns))

print("\nMissing Values:")
print(saved_inventory.isnull().sum())

print("\nDuplicate Rows:", saved_inventory.duplicated().sum())

print("\nNegative Recommended Inventory:",
      (saved_inventory["recommended_inventory"] < 0).sum())

print("\nInventory Priority Distribution:")
print(saved_inventory["inventory_priority"].value_counts())

print("\nTop 10 Inventory Requirements:")
display(
    saved_inventory[
        [
            "state",
            "vehicle_group",
            "average_monthly_demand",
            "safety_stock",
            "reorder_point",
            "recommended_inventory",
            "inventory_priority"
        ]
    ].head(10)
)

print("\n" + "=" * 70)
print("VALIDATION COMPLETED")
print("=" * 70)
```
